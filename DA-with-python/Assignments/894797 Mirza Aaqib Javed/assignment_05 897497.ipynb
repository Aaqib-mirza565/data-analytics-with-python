```python
students = {
    "name":"",
    "age":"",
    "course":"",
    "marks": []
    
}

for key in students:
    if key != "marks":
        value = input(f"Enter the value  for {key}")
        students.update({key: value})

no_of_subjects = int(input("Enter the no of subjects "))
for n in range(1,no_of_subjects + 1 ):
    
            
    val = int(input(f"Enter the marks for subject {n}"))
    students["marks"].append(val)
            
print(students)

for key, value in students.items():
    print(f"Key: {key} -> Value: {value}")
average = 0
total = 0    
for key in students:
    if key == "marks":
        x = students.get("marks")
        for i in x:
            total += i
            
        average = total / len(x)    
         
print("average of marks is ",average)            
    
for key in students:
    if key == "marks":
        x = students.get("marks")
        for i in x:
            
            if i >= 50:
                print(f" Student having marks {i} is pass ")
            else:
                print(f"student  having marks {i} is fail")    
```

```python
employees = []
employes_dic = {
    
}

emplyee_name = "name"
emplyee_sallery = 0
emplyee_department = "department"
while True:
    try:
        emplyee_name = input("Enter the name of employee want to lsit in" )
        emplyee_name = str(emplyee_name)
    except:
        print("enter the name is string not in other format")    
    if emplyee_name == "":
        break
    
    

    try:
        emplyee_sallery = input(f"Enter the sallery of {emplyee_name} " )
        emplyee_sallery = int(emplyee_sallery)
    except :
        print("Enter the integers for sallery")
    
    if emplyee_sallery == "":
        break
    
    
    try:
        
        emplyee_department = input(f"Enter the department of {emplyee_name} of employee " )
        emplyee_department = str(emplyee_department)
    except:
        print("enter the name is string not in other format")    
    if emplyee_department == "":
        break
    employees.append([emplyee_name,emplyee_sallery,emplyee_department])
    




for index, employees in enumerate(employees):
    emp_id = f"EMP-{index + 1}"
    
    
    employes_dic[emp_id] = {
        "name": employees[0],
        "salary": employees[1],
        "department": employees[2]
    }

for i in employes_dic:
    print(employes_dic.get(i))
    
highest_salary = 0
highest_emp_id = ""

for emp_id in employes_dic:
    
    current_sallery = employes_dic[emp_id]["salary"]
    
    if current_sallery > highest_salary:
        highest_salary = current_sallery
        highest_emp_id = emp_id


print("\nHighest Salary Employee Details:")
print("ID:", highest_emp_id)
print("Name:", employes_dic[highest_emp_id]["name"])
print("Salary:", employes_dic[highest_emp_id]["salary"])
print("Department:", employes_dic[highest_emp_id]["department"])

total = 0

for emp_id in employes_dic:
    sallery = employes_dic[emp_id]["salary"]
    total += sallery

print("total expence of salleries ", total)    



    
    
```
