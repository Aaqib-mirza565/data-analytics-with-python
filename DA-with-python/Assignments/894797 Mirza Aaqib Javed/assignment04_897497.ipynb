```python
person = {
    
     "name" : "aaqib",
        "age"  : "32 years",
     "cource"  : " Data Analysit",
     "city"    : "Faisalabad"
    
    
    
}

for key in person.keys():
    
    print(key)

for value in person.values():
    print(value)    
```

count frequency 


```python
numbers = [ 1,2,2,3,3,3,4,4,4,4]
frequency = {}
for i in numbers:
    if i in frequency:
        frequency[i] += 1
    else:
        frequency[i] = 1

print(frequency)        
        
    
    
    
```

word counter

```python
string = input("Enter the string: ")
words = string.split() 
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

```python
students = {
    
    "ali": 85,
    "sara":92,
    "ahmed":78,
    "zara": 95
    
}
my_list = list(students.values())




print(my_list)

max_val = my_list[0]


for num in my_list:
    if num > max_val:
        max_val = num

print("biggest number is:", max_val)
```
