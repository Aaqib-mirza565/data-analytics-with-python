6) Write a program to calculate the electricity bill (accept number of unit from user) according to the following criteria :
     Unit                                                     Price  
uptp 100 units                                             no charge
Next 200 units                                              Rs 5 per unit
After 200 units                                             Rs 10 per unit
(For example if input unit is 350 than total bill amount is Rs.3500
(For example if input unit is 97 than total bill amount is Rs.0
(For example if input unit is 150 than total bill amount is Rs.750

```python
def bill():
    units = int(input("Enter the number of units"))
    match units:
        case units if 100 < units <=200:
            print(f"total amount to pay is { units * 5} and your units are {units}")
        case units if units > 200: 
            print(f"total amount to pay is { units * 10} and your units are {units}")
        case _:
            print(f"your bill is zero and your units are {units}")

bill()                   
```

13) Take input of age of 3 people by user and determine oldest and youngest among them.

```python

def age_comparison(): 
    print("Enter the age of three persons")
    age = []
    
    for n in range(1,4):
        
        n = int(input(f"Enetr the age of person {n}"))
        age.append(n)
    print(f"your entered ages of three persons are : {age}")  
    oldest_one = max(age[0],age[1],age[2])
    
    youngest_one = min(age[0],age[1],age[2])
    
    age.remove(oldest_one)
    age.remove(youngest_one)
    print(f"most aged is {oldest_one} years old , youngest one is {youngest_one} years old and middle aged person is {age[-1]} years old  ")
age_comparison()      
        

    
```
