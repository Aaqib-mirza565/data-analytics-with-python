3. Write a program in which use all the operators we can use in Python

```python
# Write your code here

# 1. Arithmetic Operators
a = 10
b = 3
print("--- Arithmetic Operators ---")
print("Addition (+):", a + b)
print("Subtraction (-):", a - b)
print("Multiplication (*):", a * b)
print("Division (/):", a / b)
print("Floor Division (//):", a // b)
print("Modulus (%):", a % b)
print("Exponentiation (**):", a ** b)

# 2. Assignment Operators
print("\n--- Assignment Operators ---")
x = 5
print("Initial x:", x)
x += 3  # x = x + 3
print("After x += 3:", x)
x *= 2  # x = x * 2
print("After x *= 2:", x)

# 3. Comparison (Relational) Operators
print("\n--- Comparison Operators ---")
print("a > b:", a > b)
print("a < b:", a < b)
print("a == b:", a == b)
print("a != b:", a != b)

# 4. Logical Operators
print("\n--- Logical Operators ---")
p = True
q = False
print("p and q:", p and q)
print("p or q:", p or q)
print("not p:", not p)

# 5. Bitwise Operators
print("\n--- Bitwise Operators ---")
m = 5   # Binary: 0101
n = 3   # Binary: 0011
print("Bitwise AND (5 & 3):", m & n)
print("Bitwise OR (5 | 3):", m | n)
print("Bitwise XOR (5 ^ 3):", m ^ n)
print("Bitwise Left Shift (5 << 1):", m << 1)

# 6. Identity Operators
print("\n--- Identity Operators ---")
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1
print("list1 is list3:", list1 is list3)
print("list1 is list2:", list1 is list2)
print("list1 is not list2:", list1 is not list2)

# 7. Membership Operators
print("\n--- Membership Operators ---")
text = "Python"
print("'P' in text:", 'P' in text)
print("'x' not in text:", 'x' not in text)
```

4. Completes the following steps of small task:
    - Mention Marks of English , Islamiat and Maths out of 100 in 3 different variables
    - Mention Variable of Total Marks and assign 300 to it
    - Calculate Percentage

```python
# Write your code here
English_marks = int(input("Enter the marks of subject english"))
if English_marks >= 0 and English_marks < 100:
    print("English Mrks is :\t",English_marks)
else:
    print("Enter correct marks")
        
    
Math_marks = int(input("Enter the marks of subject Math"))
if Math_marks >= 0 and Math_marks < 100:
    print("English Mrks is :\t",Math_marks)
else:
    print("Enter correct marks")    

islamiat_marks = int(input("Enter the marks of subject Islamiat"))
if 0 <= islamiat_marks <= 100:
    print("English Mrks is :\t",islamiat_marks)
else:
    print("Enter correct marks")    

Marks = {
    "english":  English_marks,
    "math"   :  Math_marks,
    "islamiat": islamiat_marks,
    "Total_marks" : 300 
    
}

Marks_taken = sum([Marks["english"],Marks["math"],Marks["islamiat"]])
percentage_of_marks = (Marks_taken/Marks["Total_marks"])*100
print("percentage of Marks is = \t", percentage_of_marks,"%")
```
