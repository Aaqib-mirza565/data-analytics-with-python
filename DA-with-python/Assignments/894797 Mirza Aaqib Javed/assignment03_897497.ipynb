Write a program that accepts a list from user and print the alternate element of list. 

```python
my_list = []

while True:
    val = input("Enter element (or press Enter to stop): ")
    if val == "":  
        break
    my_list.append(val)


print("Alternate elements:", my_list[::2])
```

 Write a program that accepts a list from user. Your program should reverse the content of list and 
display it. Do not use reverse() method. 

```python
mylist = [1,2,3,4,5,6]
newlist = []
for i in mylist[::-1]:
    newlist.append(i)
print(newlist)    
```

. Find and display the largest number of a list without using built-in function max(). Your program 
should ask the user to input values in list from keyboard. 

```python
my_list = [120, 45, 2, 89, 34, 67]


max_val = my_list[0]


for num in my_list:
    if num > max_val:
        max_val = num

print("biggest number is:", max_val)
    
```
