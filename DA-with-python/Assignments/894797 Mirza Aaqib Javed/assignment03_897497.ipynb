Write a program that accepts a list from user and print the alternate element of list. 

```python
my_list = []

while True:
    val = input("Enter element (or press Enter to stop): ")
    if val == "":  
        break
    my_list.append(val)


print("Alternate elements:", my_list[::2])
```

```python
def Multiply_of_Matrix():
    try:
        number_of_rows_A = int(input("enter the number of rows you want in matrix A: "))
    except:
        print("Enter the valid integers for number of rows")
        return
        
    try:
        number_of_colums_A = int(input("enter the number of colums you want in matrix A: "))
    except:
        print("Enter the valid integers for number of colums")
        return              
    
    matrix_1 = []
    for x in range(number_of_rows_A):
        print(f"enter the values for colums of row {x+1}")
        row_list = []
        for y in range(number_of_colums_A):
            try:
                val = int(input(f"enter the element of colum {y} "))
            except:
                print("Enter the valid integers for rows")
                return
            row_list.append(val)
            
        matrix_1.append(row_list) 
            
    print(f"Matrix A = {matrix_1}\n")
    
    try:
        number_of_rows_B = int(input("enter the number of rows you want in matrix B: "))
    except:
        print("Enter the valid integers for number of rows")
        return
        
    try:
        number_of_colums_B = int(input("enter the number of colums you want in matrix B: "))
    except:
        print("Enter the valid integers for number of colums")
        return              
        
    matrix_2 = []
    for z in range(number_of_rows_B):
        print(f"enter the values for colums of row {z+1}")
        row_list_B = []
        for p in range(number_of_colums_B):
            try:
                val = int(input(f"enter the element of colum {p} "))
            except:
                print("Enter the valid integers for rows")
                return
            row_list_B.append(val)
                
        matrix_2.append(row_list_B) 
                
    print(f"Matrix B = {matrix_2}\n")
    
    if len(matrix_1[0]) != len(matrix_2):
        print("Multiplication is impossible.")
    else:
        result = []
        for i in range(len(matrix_1)):
            row_mul = []
            for j in range(len(matrix_2[0])):
                val = 0
                for k in range(len(matrix_2)):
                    val = val + matrix_1[i][k] * matrix_2[k][j]
                row_mul.append(val)
            result.append(row_mul)

        print("Result Matrix:")
        for row in result:
            print(row)

Multiply_of_Matrix()
```
