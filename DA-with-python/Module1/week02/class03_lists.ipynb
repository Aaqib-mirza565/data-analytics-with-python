# Append

```python
# append() adds a single value to the end of the list, in place

saylani_campuses = ['Z.A', 'Headoffice','Ibne-Adam']
saylani_campuses.append('Quetta-campus')  # adds one value at a time

print(saylani_campuses)

saylani_campuses.append('Naya-nazimabda-campus')
print(saylani_campuses)
```

# Pop

```python
saylani_campuses.pop()  # removes and returns the last element (index -1)
print(saylani_campuses)

# removed: 'Quetta-campus' (was at index 3)
```

```python
saylani_campuses.pop(2)  # removes and returns the element at index 2 ('Ibne-Adam')
print(saylani_campuses)

# removed: 'Ibne-Adam' (was at index 2)
```

```python
print(saylani_campuses)
print(saylani_campuses.index('Quetta-campus'))
```

```python
# extend() -- HW: adds all elements from another list (unlike append(), which adds a single item/list as one element)
```

```python
saylani_campuses = ['Malir-campus', 'Z.A', 'lahore-campus', 'Headoffice', 'Ibne-Adam','Z.A']
saylani_campuses[1] = 'Zaitoon Ashraf IT Park'  # lists are mutable -- direct index assignment works
print(saylani_campuses)
```

# Tuples

```python
saylani_campuses = ('Malir-campus', 'Z.A', 'lahore-campus', 'Headoffice', 'Ibne-Adam')
saylani_campuses[1] = 'Zaitoon Ashraf IT Park'  # tuples are immutable -- this raises TypeError
print(saylani_campuses)
print(type(saylani_campuses))
```
